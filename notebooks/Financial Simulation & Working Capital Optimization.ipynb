```python
import pandas as pd
import numpy as np

def run_financial_simulation_engine(input_path: str, unit_cost: float = 10.0, holding_rate: float = 0.20):
    print("[+] Initiating Module 5: Financial Simulation & Working Capital Optimization...")
    
    # 1. Load Inventory Decisions Dataset from Module 4
    df = pd.read_csv(input_path)
    
    # 2. Define Baseline Static Policy (Fixed 14-day inventory allocation)
    df['Static_ROP'] = np.ceil(df['Forecast_Sales'] * 14.0)
    
    # 3. Calculate Daily Capital Requirements ($)
    df['Static_Capital_Lockup'] = df['Static_ROP'] * unit_cost
    df['Dynamic_Capital_Lockup'] = df['Reorder_Point'] * unit_cost
    
    # 4. Calculate Daily Holding Costs ($)
    daily_holding_rate = holding_rate / 365.0
    df['Static_Holding_Cost'] = df['Static_Capital_Lockup'] * daily_holding_rate
    df['Dynamic_Holding_Cost'] = df['Dynamic_Capital_Lockup'] * daily_holding_rate
    
    # 5. Aggregate Macro Metrics
    total_static_holding = df['Static_Holding_Cost'].sum()
    total_dynamic_holding = df['Dynamic_Holding_Cost'].sum()
    
    avg_static_capital = df['Static_Capital_Lockup'].mean()
    avg_dynamic_capital = df['Dynamic_Capital_Lockup'].mean()
    
    capital_reduction_pct = ((avg_static_capital - avg_dynamic_capital) / avg_static_capital) * 100.0
    holding_savings = total_static_holding - total_dynamic_holding
    
    # 6. Display Financial Impact Summary
    print("\n==================================================================")
    print("FINANCIAL IMPACT & WORKING CAPITAL SIMULATION SUMMARY")
    print("==================================================================")
    print(f"Assumed Unit Cost: ${unit_cost:.2f} | Annual Holding Rate: {holding_rate*100:.0f}%")
    print(f"Average Capital Lockup (Static Policy) : ${avg_static_capital:,.2f} per Store/Day")
    print(f"Average Capital Lockup (Dynamic Engine): ${avg_dynamic_capital:,.2f} per Store/Day")
    print(f"--> Working Capital Optimization Reduction: {capital_reduction_pct:.2f}%")
    print("------------------------------------------------------------------")
    print(f"Total Projected Holding Cost (Static) : ${total_static_holding:,.2f}")
    print(f"Total Projected Holding Cost (Dynamic): ${total_dynamic_holding:,.2f}")
    print(f"--> Net Holding Cost Savings Generated : ${holding_savings:,.2f} ({capital_reduction_pct:.2f}% Savings)")
    print("==================================================================")
    
    # Export final simulation decision ledger
    df.to_csv("rossmann_financial_simulation.csv", index=False)
    print(f"[+] Final Financial Ledger exported to 'rossmann_financial_simulation.csv'. Total Rows: {len(df):,}")
    
    return df

# Execute Module 5 Simulation
df_financials = run_financial_simulation_engine("rossmann_inventory_decisions.csv")
```
